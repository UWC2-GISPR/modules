# Getting to Know Jupyter Notebooks

A quick tour of how notebooks work: cells, shortcuts, moving things around, Markdown vs. code, and switching kernels.

Work through this top to bottom, actually running the code cells (don't just read them) — muscle memory is the point.

## 1. The basic unit: cells

A notebook is a list of **cells**. Each cell is one of two types:

- **Code** — Python that runs in the notebook's kernel (the running Python process behind the scenes)
- **Markdown** — formatted text, like this cell

A cell has two modes:

- **Edit mode** (green border, cursor blinking) — you're typing inside the cell
- **Command mode** (blue border) — the cell is selected but you're not typing; keyboard shortcuts act on the whole cell/notebook

Press `Enter` to go from command mode into edit mode. Press `Esc` to go back to command mode.

## 2. Running a cell

- `Shift+Enter` — run the cell, select the next one (or create a new one if this is the last cell)
- `Ctrl+Enter` — run the cell, stay on it
- `Alt+Enter` (or `Option+Enter` on Mac) — run the cell, insert a new one right after

Try it on the code cell below.

````python
message = "Hello, notebook!"
print(message)
````

## 3. Cells share state

All code cells in a notebook run in the **same kernel**, so variables defined in one cell are available in the next. Run the cell below — it uses `message` from the cell above.

````python
shout = message.upper() + "!!!"
shout
````

Notice two things:

1. You didn't need `print()` — a bare expression on the last line of a cell displays its value automatically.
2. The number in `[ ]` to the left of the cell (the **execution count**) tells you the *order cells were run in*, not their position on the page. It's easy to run cells out of order and get confused — if results look wrong, try **Kernel → Restart & Run All** to run everything fresh, top to bottom.

## 4. Markdown vs. code cells

Change a cell's type from the toolbar, or with keyboard shortcuts in **command mode**:

- `Y` — convert the selected cell to a **code** cell
- `M` — convert the selected cell to a **Markdown** cell
- `R` — convert to raw (rarely used)

Markdown cells support headers, lists, bold/italic, links, code blocks, and even math.

### Markdown formatting cheat sheet

```
# H1 header          -> # H1
## H2 header         -> ## H2
*italic*             -> *italic*
**bold**             -> **bold**
`inline code`        -> `inline code`
- bullet list        -> - item
1. numbered list     -> 1. item
[a link](https://x)  -> [a link](https://example.com)
```

Rendered examples:

- **bold text**, *italic text*, `inline code`
- A numbered list:
  1. First
  2. Second
- Even LaTeX math, if you like: $e^{i\pi} + 1 = 0$

Double-click any Markdown cell to see its raw source, then `Shift+Enter` to render it again.

## 5. Adding, deleting, and moving cells

All in **command mode** (press `Esc` first if you're in edit mode):

| Shortcut | Action |
|---|---|
| `A` | insert a new cell **above** |
| `B` | insert a new cell **below** |
| `D`, `D` (press D twice) | delete the selected cell |
| `Z` | undo cell deletion |
| `X` | cut the selected cell |
| `C` | copy the selected cell |
| `V` | paste cell below |
| `Ctrl+Shift+Alt+[` / `]` (VS Code) or drag | move a cell up/down |

In VS Code's notebook UI specifically, you can also grab the `⋮⋮` handle on the left of a cell and drag it, or use the up/down arrow icons that appear when you hover over a cell.

**Try it:** select this cell, press `B` to insert a new cell below it, turn it into a code cell with `Y`, and write something in it.

## 6. Other handy shortcuts (command mode)

| Shortcut | Action |
|---|---|
| `Enter` | enter edit mode |
| `Esc` | enter command mode |
| `Shift+Enter` | run cell, move to next |
| `Ctrl+Enter` | run cell, stay put |
| `Alt+Enter` | run cell, insert new below |
| `Ctrl+/` | comment/uncomment code (in edit mode) |
| `Ctrl+Shift+P` | command palette (find any notebook action) |
| `L` | toggle line numbers on a code cell |
| `Ctrl+S` | save the notebook |

When in doubt, `Ctrl+Shift+P` and search — it lists every notebook command with its current shortcut.

## 7. What is a kernel, and switching it

The **kernel** is the actual running process (e.g. a Python interpreter) that executes your code cells and holds all your variables in memory. A notebook file doesn't have a kernel baked in — it just records which one it *expects*.

The cell below asks the running kernel a couple of questions about itself.

````python
import sys
print("Python executable:", sys.executable)
print("Python version:", sys.version)
````

To switch which environment/interpreter runs your code (e.g. swap from a base Python install to a project's virtual environment or conda env):

- **VS Code**: click the kernel name in the top-right of the notebook toolbar and pick a different one from the list, or `Ctrl+Shift+P` -> "Notebook: Select Kernel"
- **Classic Jupyter (browser)**: menu **Kernel -> Change Kernel**

Switching kernels **restarts** the Python process, so all your variables are wiped — you'll need to re-run cells from the top. This is also the fix when a notebook gets into a confusing state: **Kernel -> Restart** clears everything and gives you a clean slate; **Restart & Run All** additionally re-executes every cell in order.

## 8. Magic commands

Lines starting with `%` (line magic) or `%%` (cell magic) are Jupyter-specific shortcuts, not Python — handled by the kernel, not the language.

````python
# list every variable currently defined in this kernel
%who
````

````python
# time how long a line takes to run
%timeit sum(range(10_000))
````

````python
# a leading ! runs a shell command instead of Python
!python --version
````

## 9. Recap

- Cells are **code** or **Markdown**; toggle with `Y` / `M` in command mode
- `Shift+Enter` runs a cell; execution order (not page order) determines what state you're in
- `A`/`B` insert cells above/below, `D D` deletes, `X`/`C`/`V` cut/copy/paste
- The **kernel** holds your running state; switch it from the toolbar (VS Code) or Kernel menu (classic Jupyter); restarting wipes variables
- `%magic` commands and `!shell` commands extend what a code cell can do beyond plain Python
